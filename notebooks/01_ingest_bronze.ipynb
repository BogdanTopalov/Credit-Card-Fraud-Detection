```python
from pyspark.sql import functions as F


TEST_FRAUD_CSV = "Files/fraud_test.csv"
TRAIN_FRAUD_CSV = "Files/fraud_train.csv"


def ingest_raw_data(csv_path, table_name):
    df = spark.read.format("csv").option("header","true").load(csv_path)

    total_rows = df.count()
    unique_rows = df.dropDuplicates().count()

    print('fraud_test.cvs')
    print(f"Total rows: {total_rows:,d}")
    print(f"Unique rows: {unique_rows:,d}")
    print(f"Duplicate rows: {total_rows - unique_rows:,d}")


    # Column counts
    for c in df.columns:
        unique_count = df.select(F.countDistinct(c)).collect()[0][0]
        
        null_count = df.select(F.sum(F.col(c).isNull().cast("int"))).collect()[0][0]
        
        print(f'{c}: unique values = {unique_count:,d} | null values = {null_count:,d}')


    # Show fraud imbalance
    df.groupBy("is_fraud").count().show()

    # Write raw data
    df.write.format("delta").mode("overwrite").saveAsTable(table_name)

    display(df)
```

### Ingest Test Fraud Data

```python
ingest_raw_data(TEST_FRAUD_CSV, 'raw_test_transactions')
```

### Ingest Train Fraud Data

```python
ingest_raw_data(TRAIN_FRAUD_CSV, 'raw_train_transactions')
```
