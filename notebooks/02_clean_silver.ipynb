```python
from pyspark.sql.functions import col, to_timestamp, to_date


def clean_raw_data(raw_table_name, clean_table_name):
    df = spark.sql(f"SELECT * FROM fraud_detection_lakehouse.dbo.{raw_table_name}")

    # Remove index column
    df = df.drop('_c0')

    # Change non-string column types
    df = (
        df
        .withColumn("trans_date_trans_time", to_timestamp("trans_date_trans_time"))
        .withColumn("amt", col("amt").cast("double"))
        .withColumn("zip", col("zip").cast("int"))
        .withColumn("lat", col("lat").cast("double"))
        .withColumn("long", col("long").cast("double"))
        .withColumn("city_pop", col("city_pop").cast("int"))
        .withColumn("dob", to_date("dob"))
        .withColumn("unix_time", col("unix_time").cast("long"))
        .withColumn("merch_lat", col("merch_lat").cast("double"))
        .withColumn("merch_long", col("merch_long").cast("double"))
        .withColumn("is_fraud", col("is_fraud").cast("int"))
    )

    # Remove duplicated rows and rows with missing data
    df = df.dropDuplicates().dropna()

    # Write clean data table
    df.write.format("delta").mode("overwrite").saveAsTable(clean_table_name)

    df.printSchema()
```

### Clean Test Fraud table

```python
clean_raw_data('raw_test_transactions', 'cleaned_test_transactions')
```

### Clean Train Fraud table

```python
clean_raw_data('raw_train_transactions', 'cleaned_train_transactions')
```
